```python
# First let's import the packages we will use in this project
# You can do this all now or as you need them
import pandas as pd
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt
import matplotlib.mlab as mlab
import matplotlib
plt.style.use('ggplot')
from matplotlib.pyplot import figure

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12,8)

pd.options.mode.chained_assignment = None

#df = pd.read_csv(r"C:\Users\chloe\Downloads\movie\movies.csv")
df = pd.read_csv("movies.csv")
df
```

```python
# looping through the data to see if anything missing

for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print('{} - {}%'.format(col, round(pct_missing*100)))
```

```python
#dropping columns with null values
df = df.dropna()
df
```

```python
print(df.dtypes)
```

```python
# Changing budget and gross into intergers
df['budget']=df['budget'].astype('int64')
df['gross']=df['gross'].astype('int64')
```

```python
df
```

```python
#Making sure data has correct years
df['yearcorrect'] = df['released'].astype(str).str.split().str[2]
df
```

```python
#displaying all rows for better analysis
pd.set_option('display.max_rows',None)
```

```python
#droping duplicate values
df['company'].drop_duplicates().sort_values(ascending=False)
```

```python
df.sort_values(by=['gross'], inplace=False, ascending=False)
```

```python
#Budget vs gross earnings scatter plot
plt.scatter(x=df['budget'], y=df['gross'], alpha=0.5)
plt.title('Budget vs Gross Earnings')
plt.xlabel('Gross Earnings')
plt.ylabel('Budget for Film')
plt.show()
```

```python
sns.regplot(x="gross", y="budget", data=df)
```

```python
#pearson, kendall, spearman correlation
df.corr(method='pearson', numeric_only=True) 
```

```python
#correlation matrix
correlation_matrix = df.corr(numeric_only=True)

sns.heatmap(correlation_matrix, annot = True)

plt.title("Correlation matrix for Numeric Features")

plt.xlabel("Movie features")

plt.ylabel("Movie features")

plt.show()
```

```python
#Assigning random numbers to text
df_numerized = df.copy()

for col_name in df_numerized.columns:
    if(df_numerized[col_name].dtype == 'object'):
        df_numerized[col_name] = df_numerized[col_name].astype('category')
        df_numerized[col_name] = df_numerized[col_name].cat.codes

df_numerized
```

```python
#calculating mean gross value
company_gross = df.groupby('company')['gross'].mean()
```

```python
#Ranking companies
company_gross.sort_values(ascending=False)
```

```python
#Top 10 companies
company_stats = (df.groupby('company')['gross'].agg(['mean', 'count']).reset_index())

company_stats = company_stats.rename(columns={'mean': 'mean_gross','count': 'movie_count'})

company_stats = company_stats.sort_values(by='mean_gross',ascending=False)

company_stats['company_rank'] = company_stats['mean_gross'].rank(ascending=False, method='min')

company_stats.head(10)
```

```python
#Top 10 Mean Companies mean gross revenue
top_10 = company_stats.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(data=top_10, x='mean_gross', y='company', hue='company', legend=False)

plt.title('Top 10 Production Companies by Mean Gross Revenue')
plt.xlabel('Mean Gross Revenue')
plt.ylabel('Production Company')

plt.tight_layout()
plt.show()
```

```python
# Pearson correlation
correlation = company_stats[
    ['company_rank', 'mean_gross']
].corr(method='pearson')

print(correlation)
```

```python
#SPearman
spearman_corr = company_stats[
    ['company_rank', 'mean_gross']
].corr(method='spearman')

print(spearman_corr)
```

```python
#Filtering companies with less then 5 movies
company_stats_filtered = company_stats[
    company_stats['movie_count'] >= 5
].sort_values(
    by='mean_gross',
    ascending=False
)

company_stats_filtered.head(10)
```

```python
#Plotting the results
plt.figure(figsize=(12, 6))

sns.barplot(
    data=company_stats_filtered.head(10),
    x='mean_gross',
    y='company',
    hue='company',
    legend=False
)

plt.title('Top Production Companies by Mean Gross Revenue')
plt.xlabel('Mean Gross Revenue')
plt.ylabel('Production Company')

plt.tight_layout()
plt.show()
```
